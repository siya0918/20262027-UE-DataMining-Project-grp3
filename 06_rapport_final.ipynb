## 1. Notre méthode

On a travaillé en trois temps.

**1. Explorer chaque domaine séparément.** Chaque domaine (santé, écologie, public records, logistique) a d'abord été traité seul, en deux notebooks :
- un notebook de **nettoyage** : doublons, formats des nombres et des dates, catégories mal écrites, valeurs manquantes ;
- un notebook d'**analyse** : on décrit les données, on cherche ce qui sort de l'ordinaire avec les méthodes du cours, et on garde aussi les pistes qui ne donnent rien.

**2. Noter les observations et les questions.** À la fin de chaque analyse, on a noté ce qu'on avait trouvé et les questions restées sans réponse. Le fichier `pistes.md` garde la trace de ce raisonnement.

**3. Faire circuler les questions entre les domaines.** Souvent, un indice trouvé dans un domaine posait une question à laquelle seul un autre domaine pouvait répondre. Par exemple :
- en écologie, un rapport parle d'une équipe entrée dans une grotte, sans dire laquelle → on a cherché qui dans public records ;
- en public records, cette équipe emporte du matériel au froid → on a cherché où il allait dans la logistique ;
- en logistique, les livraisons arrivent dans la clinique où des dirigeants reçoivent un traitement secret repéré en santé.

Le notebook `05_croisements` répond enfin aux questions qui demandent de relier plusieurs domaines à la fois.

Le narratif s'est donc construit pas à pas, chaque domaine apportant une partie de l'histoire.

## 2. Ce que chaque domaine nous a appris

### Santé (`01_health`)
- **Question de départ** : y a-t-il des patients ou des actes médicaux qui sortent de l'ordinaire ?
- **Ce qu'on a fait** : une ligne par patient, l'IQR sur chaque mesure, le LOF puis l'Isolation Forest pour trouver les profils atypiques, une ACP pour les visualiser, l'autocorrélation sur les signaux cardiaques, le TF-IDF sur les notes des médecins, et des règles d'association sur les actes.
- **Ce qu'on a trouvé** :
  - 11 patients au corps inhabituel (yeux, démarche, poitrine, deux rythmes cardiaques), que les caméras reconnaissent mal ;
  - 8 dirigeants qui reçoivent plusieurs fois un acte secret, MRP-42, et paraissent 11 ans plus jeunes que leur âge ;
  - une note de médecin qui relie un patient atypique à un nom public : Elias Voss.
- **Pistes sans résultat** : des erreurs d'appareil (toutes les mesures sont de bonne qualité), le LOF (qui ne voit pas un petit groupe de patients qui se ressemblent).
- **Questions transmises** : qui est Elias Voss ? Qui sont ces dirigeants ? De quoi est fait MRP-42 ?

### Écologie (`02_ecology`)
- **Question de départ** : qu'est-ce qui fait varier le nombre de chauves-souris ?
- **Ce qu'on a fait** : statistiques descriptives et boxplots, autocorrélation (cycle d'un an), corrélations de Pearson et de Spearman, K-means, IQR et z-score modifié, règles d'association, et TF-IDF sur les rapports de terrain.
- **Ce qu'on a trouvé** :
  - le nombre de chauves-souris suit les saisons, et baisse avec la perturbation et les pesticides ;
  - les chauves-souris aident les durians à donner des fruits ;
  - près d'un quart des fortes baisses n'ont aucune explication ordinaire ;
  - un seul rapport de terrain ne ressemble à aucun autre : il décrit une baisse de chauves-souris à KNT-KES après l'entrée d'une équipe autorisée, avec des liens coupés.
- **Pistes sans résultat** : le regroupement des sites, l'IQR sur les comptages bruts, les grottes à accès contrôlé.
- **Question transmise** : qui est entré dans la grotte ?

### Public records (`03_public_records`)
- **Questions de départ** : celles de la santé et de l'écologie, puis ce que contiennent ces registres.
- **Ce qu'on a fait** : IQR et z-score modifié sur les durées de mandat, règles d'association sur les demandes d'accès, process mining (matrice de transition) sur les plaintes, graphes et Louvain, ACP, CAH, DBSCAN et LOF sur les dirigeants, TF-IDF et LSA sur les documents et les mémos, et comparaison de noms.
- **Ce qu'on a trouvé** :
  - c'est **Horizon** qui est entrée à KNT-KES, de nuit, avec du matériel de stockage à froid ; elle est la seule à apporter ce matériel ;
  - **Elias Voss** est le président de **Meridian BioLogistics**, dont les mémos parlent d'un code secret « CHAMELEO » et d'unités biologiques qui disparaissent ;
  - les 8 dirigeants traités par MRP-42 sont en poste depuis 39 à 57 ans ;
  - des personnes ont un âge impossible (nées en 1862 ou en 1900, et toujours actives) ;
  - 6 patients atypiques ont un nom presque identique à une personne de public records.
- **Questions transmises** : où va ce qu'Horizon emporte ? Que transportent les livraisons citées dans les plaintes ?

### Logistique (`04_logistics`)
- **Questions de départ** : celles de public records et de la santé.
- **Ce qu'on a fait** : statistiques descriptives, IQR, z-score, DBSCAN et LOF sur les stocks, ACP et CAH, graphe des expéditions, moyenne glissante, puis règles d'association sur les expéditions, et comparaison des dates avec les autres domaines.
- **Ce qu'on a trouvé** :
  - 6 heures après chaque visite nocturne dans une grotte, Horizon envoie des échantillons biologiques au froid à Meridian (18 fois sur 18) ;
  - Meridian livre la clinique Asterion, et chaque livraison est suivie 8 jours plus tard d'un MRP-42 donné à un dirigeant (26 fois sur 26) ;
  - le jour de chaque MRP-42, du matériel est retiré du stock d'Asterion « pour consommation » (30 fois sur 30).

### Croisements (`05_croisements`)
- **Questions de départ** : celles qui demandent de relier plusieurs domaines.
- **Ce qu'on a fait** : jointures dans le temps entre domaines, comparaison à la normale saisonnière, similarité de noms, recherche de mots-clés dans les documents.
- **Ce qu'on a trouvé** :
  - les chauves-souris baissent après presque chaque visite d'Horizon (13 fois sur 14) ;
  - parmi les 240 patients, seuls les patients atypiques ont un quasi-homonyme dans public records (même nom à une lettre près, même naissance, même région) ;
  - deux lignées de dirigeants se succèdent depuis le XIXe siècle, avec des noms qui se ressemblent ;
  - des documents décrivent les patients atypiques comme une population non humaine (« Chamaeleonidian ») et parlent de chauves-souris « consommées ».

## 3. Les méthodes du cours utilisées

| Méthode | Où | À quoi elle a servi |
|---|---|---|
| Nettoyage (doublons, formats, valeurs manquantes) | Tous les notebooks de nettoyage | Rendre les données fiables avant l'analyse |
| Statistiques descriptives, boxplots | Tous | Décrire ce qui est normal |
| IQR, z-score modifié | Santé, écologie, public records, logistique | Repérer les valeurs hors norme |
| LOF, Isolation Forest, DBSCAN | Santé, public records, logistique | Repérer les individus atypiques sur plusieurs mesures |
| Standardisation, ACP | Santé, public records, logistique | Mettre les mesures à la même échelle, visualiser les profils |
| K-means, CAH, score de silhouette | Écologie, public records, logistique | Chercher des groupes de sites, de dirigeants ou de stocks |
| Corrélations de Pearson et de Spearman | Écologie, public records, logistique | Mesurer les liens entre variables |
| Règles d'association (support, confiance, lift) | Santé, écologie, public records, logistique | Vérifier que des liens ne sont pas dus au hasard |
| Graphes, centralité, Louvain | Public records, logistique | Trouver les acteurs centraux et les groupes |
| Process mining (matrice de transition) | Public records | Suivre le parcours des plaintes |
| Séries temporelles, autocorrélation, moyenne glissante | Écologie, santé, logistique | Trouver les cycles saisonniers et les rythmes cardiaques |
| TF-IDF, similarité cosinus, LSA | Santé, écologie, public records | Trouver les textes qui ne ressemblent à aucun autre |

## 4. Le narratif

**Ce que les données établissent**
1. Horizon entre de nuit dans des grottes ; après presque chaque visite, les chauves-souris baissent fortement.
2. 6 heures après chaque visite, Horizon envoie des échantillons biologiques gardés au froid à Meridian BioLogistics.
3. Meridian livre la clinique Asterion ; 8 jours plus tard, du matériel est retiré du stock et un dirigeant reçoit MRP-42.
4. Ces dirigeants paraissent 11 ans plus jeunes que leur âge et sont au pouvoir depuis 40 à 50 ans.
5. Les patients atypiques vivent sous des noms légèrement modifiés, et deux lignées de dirigeants se remplacent depuis le XIXe siècle.

**Ce que les documents décrivent**
- Les patients atypiques forment une population cachée, non humaine (« Chamaeleonidian »), qui change d'identité au fil des siècles ; Elias Voss, président de Meridian, en ferait partie.
- Les chauves-souris collectées sont « consommées », et une partie des envois part vers une destination non terrestre.
- « Unwabu », l'autorité citée dans les mémos, est le nom zoulou du caméléon : dans le mythe, c'est le caméléon qui devait apporter l'immortalité aux hommes. Ça rejoint le code CHAMELEO et les traits des patients atypiques (yeux indépendants, démarche qui se balance, mal vus par les caméras).

**Notre conclusion**
Une population cachée, non humaine et proche du caméléon selon les documents, fait capturer des chauves-souris *Eonycteris spelaea* dans les grottes. Elle s'en sert pour préparer MRP-42, un traitement qui garde jeunes et au pouvoir des dirigeants humains.

## 5. Limites

- On ne voit jamais le contenu réel des colis, seulement leur catégorie déclarée, qui peut être fausse.
- La composition de MRP-42 n'est écrite nulle part : le lien avec les chauves-souris repose sur des dates qui se répètent sans exception, et sur un témoignage.
- La nature non humaine des patients atypiques vient de documents, pas de mesures.
- Certaines pistes reposent sur peu de cas (8 dirigeants, 14 visites).
- Le sens d'« Unwabu » (caméléon) vient d'une recherche extérieure sur la mythologie zouloue, pas des données.
